```python

def format_date(date_str):
    """
    Converts a string 'YYYY-MM-DD' to ISO format.
    """
    dt = datetime.datetime.strptime(date_str, '%Y-%m-%d')
    return dt.strftime('%Y-%m-%dT00:00:00.000-00:00')
```

```python
# import the FuelEU penalties dataset
fueleu_penalties_df = pd.read_excel('./data/raw/FuelEU_MGO_penalty.xlsx')
fueleu_penalties_df
```

```python
fueleu_penalties_df = fueleu_penalties_df.iloc[1:,[0,2]]
fueleu_penalties_df.columns = ['Date', 'FuelEU_MGO_penalty']
fueleu_penalties_df.reset_index(drop=True, inplace=True)
fueleu_penalties_df
```

```python
for df in [renewables_df, brent_df, lsmgo_df, eur_usd_df, fed_funds_df, fueleu_penalties_df]:
    df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
    df['Date'] = df['Date'].dt.normalize()
```

```python
final_df = renewables_df.merge(brent_df, on='Date', how='outer')
final_df = final_df.merge(lsmgo_df, on='Date', how='outer')
final_df = final_df.merge(eur_usd_df, on='Date', how='outer')
final_df = final_df.merge(fed_funds_df, on='Date', how='outer')
final_df = final_df.merge(fueleu_penalties_df, on='Date', how='outer')
```

```python
final_df = final_df[(final_df['Date'] >= '2021-04-26')&(final_df['Date'] <= '2026-04-23')]
final_df.reset_index(drop=True, inplace=True)
```

```python
final_df
```

```python
#drop unused columns
final_df = final_df.drop(columns=['UCO_fob_ARA', 'UCO_(hydrotreating)_fob_ARA', 'UCO_cif_ARA'])
```

```python
final_df
```

```python
#Interpolates missing values
final_df['Date'] = pd.to_datetime(final_df['Date'])

final_df = final_df.set_index('Date')

final_df = final_df.asfreq('D')

final_df = final_df.reset_index()

final_df[final_df.columns.difference(['Date'])] = final_df[final_df.columns.difference(['Date'])].apply(pd.to_numeric, errors='coerce')

final_df = final_df.interpolate(method='linear')
final_df.fillna(0, inplace=True)
final_df
```

```python
final_df.to_csv('final_dataset.csv', index=False)
```
